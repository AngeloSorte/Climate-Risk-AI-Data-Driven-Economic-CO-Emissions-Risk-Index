```python
# =========================================
# 🌍 CLIMATE RISK AI
# REAL DATA + INTERPRETABLE INDEX
# =========================================

!pip install pandas numpy matplotlib seaborn scikit-learn requests --quiet

import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

# =========================================
# 1. COUNTRIES
# =========================================

countries = ["USA","CHN","IND","DEU","ITA","FRA","BRA","ZAF","JPN","CAN"]

# =========================================
# 2. GDP (WORLD BANK)
# =========================================

def get_gdp(country):
    url = f"https://api.worldbank.org/v2/country/{country}/indicator/NY.GDP.MKTP.CD?format=json&per_page=20000"
    r = requests.get(url).json()

    if not isinstance(r, list) or len(r) < 2:
        return pd.DataFrame()

    data = r[1]

    rows = []
    for d in data:
        if d["value"] is not None:
            rows.append({
                "country": country,
                "year": int(d["date"]),
                "gdp": float(d["value"])
            })

    return pd.DataFrame(rows)

gdp = pd.concat([get_gdp(c) for c in countries], ignore_index=True)

# clean
gdp = gdp.dropna()

# =========================================
# 3. CO2 (OWID - GLOBAL CARBON PROJECT)
# =========================================

co2 = pd.read_csv(
    "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
)

co2 = co2[co2["iso_code"].isin(countries)]
co2 = co2[["iso_code","year","co2","co2_per_capita"]]
co2 = co2.rename(columns={"iso_code":"country"})

# =========================================
# 4. MERGE
# =========================================

df = pd.merge(gdp, co2, on=["country","year"], how="inner")
df = df.dropna()
df = df.sort_values(["country","year"])

# =========================================
# 5. FEATURE ENGINEERING (REAL RISK FACTORS)
# =========================================

# growth rates
df["gdp_growth"] = df.groupby("country")["gdp"].pct_change()
df["co2_growth"] = df.groupby("country")["co2"].pct_change()

# carbon intensity
df["carbon_intensity"] = df["co2"] / df["gdp"]

df = df.dropna()

# =========================================
# 6. NORMALIZATION (IMPORTANT FIX)
# =========================================

features = ["co2_per_capita", "co2_growth", "carbon_intensity", "gdp_growth"]

X = df.groupby("country")[features].mean()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_norm = pd.DataFrame(X_scaled, index=X.index, columns=features)

# =========================================
# 7. CLIMATE RISK INDEX (REAL MODEL)
# =========================================

# weighted index (interpretable)
risk_score = (
    0.35 * X_norm["co2_per_capita"] +
    0.25 * X_norm["co2_growth"] +
    0.25 * X_norm["carbon_intensity"] -
    0.15 * X_norm["gdp_growth"]   # growth reduces risk
)

risk_score = (risk_score - risk_score.min()) / (risk_score.max() - risk_score.min()) * 100

risk_df = pd.DataFrame({
    "risk_score": risk_score
}).sort_values("risk_score", ascending=False)

# classification
def label(x):
    if x > 66:
        return "HIGH"
    elif x > 33:
        return "MEDIUM"
    else:
        return "LOW"

risk_df["risk_level"] = risk_df["risk_score"].apply(label)

print("\n🌍 CLIMATE RISK INDEX (0–100):")
print(risk_df)

# =========================================
# 8. VISUALIZATION (CLEAN + CORRECT)
# =========================================

plt.figure(figsize=(8,5))

colors = risk_df["risk_level"].map({
    "HIGH":"red",
    "MEDIUM":"orange",
    "LOW":"green"
})

plt.bar(risk_df.index, risk_df["risk_score"], color=colors)

plt.title("Climate Risk Index by Country")
plt.ylabel("Risk Score (0–100)")
plt.xticks(rotation=45)
plt.show()

# =========================================
# 9. FINAL INSIGHT
# =========================================

print("\n🔴 HIGH RISK:")
print(risk_df[risk_df["risk_level"]=="HIGH"])

print("\n🟡 MEDIUM RISK:")
print(risk_df[risk_df["risk_level"]=="MEDIUM"])

print("\n🟢 LOW RISK:")
print(risk_df[risk_df["risk_level"]=="LOW"])
```
